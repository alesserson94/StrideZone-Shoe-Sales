```python
#Import pandas & datetime packages
import pandas as pd
import datetime as dt

#Display all rows of data
pd.set_option('display.max_rows', None)
```

```python
#Load CSV and convert to DataFrame
shoe_sales=pd.read_csv('/Users/aaronless/Desktop/Data Analysis Porfolio Projects/Marketing:Sales/Shoe Sales/Original CSVs/sales_transactions.csv')
shoe_sales.head(50)
```

```python
#Clean date column to ensure all values are of consistent formatting
shoe_sales['order_datetime']=pd.to_datetime(shoe_sales['order_datetime'],errors='coerce')
shoe_sales['order_datetime'].value_counts().head(50)
```

```python
shoe_sales['category'].value_counts()

#Clean category column to ensure consistent text formatting
shoe_sales['category']=shoe_sales['category'].str.title()
shoe_sales['category'].value_counts()

#Replace typos/abbreviations where appropriate
shoe_sales['category']=shoe_sales['category'].replace({'Bball':'Basketball','Runing':'Running'})
shoe_sales['category'].value_counts()
```

```python
shoe_sales['gender'].value_counts()

#Standardize text formatting
shoe_sales['gender']=shoe_sales['gender'].str.title()
shoe_sales['gender'].value_counts()

women_mapping={"Women":"Womens","W":"Womens"}
shoe_sales['gender'].map(women_mapping)
shoe_sales['gender'].value_counts()
```

```python
#Clean shoe size column by removing text characters and trailing whitespaces
shoe_sales['size'].value_counts()

shoe_sales['size']=shoe_sales['size'].str.strip('US')
shoe_sales['size'].value_counts()

shoe_sales['size']=shoe_sales['size'].str.strip('ize')
shoe_sales['size'].value_counts()

shoe_sales['size']=shoe_sales['size'].str.strip('.0')
shoe_sales['size'].value_counts()

shoe_sales['size']=shoe_sales['size'].str.strip()
shoe_sales['size'].value_counts()
```

```python
#Clean order_status column to ensure consistent text formatting
shoe_sales['order_status'].value_counts()

shoe_sales['order_status']=shoe_sales['order_status'].str.title()
shoe_sales['order_status'].value_counts()

shoe_sales['order_status']=shoe_sales['order_status'].replace({'Complete':'Completed','Return':'Returned','Exchange':'Exchanged','Canceled':'Cancelled'})
shoe_sales['order_status'].value_counts()
```

```python
#Standardize formatting across sales channel column
shoe_sales['channel'].value_counts()

shoe_sales['channel']=shoe_sales['channel'].replace({'In Sto':'In-Store','Insto':'In-Store','In-Sto':'In-Store'})
shoe_sales['channel'].value_counts()

shoe_sales['channel']=shoe_sales['channel'].str.title()
shoe_sales['channel'].value_counts()

shoe_sales['channel']=shoe_sales['channel'].replace({'Phone Order':'Phone','In Store':'In-Store','Instore':'In-Store','Web':'Online'})
shoe_sales['channel'].value_counts()
```

```python
#Ensure consistent formatting for payment method column
shoe_sales['payment_method'].value_counts()

shoe_sales['payment_method']=shoe_sales['payment_method'].str.title()
shoe_sales['payment_method'].value_counts()

shoe_sales['payment_method']=shoe_sales['payment_method'].replace({'Paypal':'PayPal','Debit':'Debit Card','Applepay':'Apple Pay','Cc':'Credit Card','Gc':'Gift Card','Credit':'Credit Card','Pay Pal':'PayPal','Gift-Card':'Gift Card'})
shoe_sales['payment_method'].value_counts()
```

```python
#Clean unit price column to ensure all decimal values with no $ attached 
shoe_sales['unit_price']=shoe_sales['unit_price'].str.replace('$','')
shoe_sales['unit_price'].value_counts()

shoe_sales['unit_price']=shoe_sales['unit_price'].str.replace('-','')
shoe_sales['unit_price'].value_counts().head(50)
```

```python
#Ensure all quantity values are positive integers, no negative values
shoe_sales['quantity'].value_counts()

shoe_sales['quantity']=shoe_sales['quantity'].replace(-1,1)
shoe_sales['quantity'].value_counts()
```

```python
#Remove % sign for discount column
shoe_sales['discount_pct'].value_counts()

shoe_sales['discount_pct']=shoe_sales['discount_pct'].str.strip('%')
shoe_sales['discount_pct'].value_counts()
```
